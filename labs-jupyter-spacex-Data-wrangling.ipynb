# SpaceX Falcon 9 – Data Wrangling
Explore the launch data and create the landing outcome label **Class** (1 = first stage landed, 0 = did not land).

```python
import pandas as pd
import numpy as np
df = pd.read_csv("data/dataset_part_1.csv")
df.head(10)
```

```python
df.isnull().sum() / len(df) * 100
```

```python
df.dtypes
```

## Task 1: Calculate the number of launches on each site

```python
df['LaunchSite'].value_counts()
```

## Task 2: Calculate the number and occurrence of each orbit

```python
df['Orbit'].value_counts()
```

## Task 3: Calculate the number and occurrence of mission outcomes

```python
landing_outcomes = df['Outcome'].value_counts()
landing_outcomes
```

```python
for i, outcome in enumerate(landing_outcomes.keys()):
    print(i, outcome)
```

```python
# outcomes where the first stage did not land successfully
bad_outcomes = {o for o in landing_outcomes.keys() if o.startswith('False') or o.startswith('None')}
bad_outcomes
```

## Task 4: Create a landing outcome label from the Outcome column

```python
landing_class = [0 if outcome in bad_outcomes else 1 for outcome in df['Outcome']]
df['Class'] = landing_class
df[['Class']].head(8)
```

```python
df.head(5)
```

```python
print("Success rate:", df["Class"].mean())
```
